```sql

SELECT
  st.city,
  COUNT_IF(s.status = 'COMPLETED') AS completed_sessions,
  SUM(CASE WHEN s.status = 'COMPLETED' THEN s.energy_kwh ELSE 0 END) AS energy_kwh,
  round(SUM(CASE WHEN s.status = 'COMPLETED' THEN s.billed_amount ELSE 0 END),2) AS total_billed,
  ROUND(100.0 * COUNT_IF(s.status = 'COMPLETED') / NULLIF(COUNT(*), 0), 2) AS success_rate_pct
FROM gold.fact_sessions AS s
JOIN gold.dim_stations AS st
  ON s.station_id = st.station_id
JOIN gold.dim_date AS d
  ON s.date_key = d.date_key
WHERE st.city = initcap(:city)
  AND d.calendar_date >= to_date(:start_date, 'dd-MM-yyyy')
  AND d.calendar_date <= to_date(:end_date, 'dd-MM-yyyy')
GROUP BY st.city
ORDER BY st.city;
```
