```sql
WITH charger_stats AS (
  SELECT
    charger_id,
    COUNT_IF(status = 'COMPLETED') AS completed_sessions,
    COUNT_IF(status = 'FAILED') AS failed_sessions
  FROM gold.fact_sessions
  GROUP BY charger_id
)
SELECT
  COUNT(*) AS breach_count
FROM charger_stats
WHERE completed_sessions + failed_sessions >= 10
  AND ROUND(100.0 * failed_sessions / (completed_sessions + failed_sessions), 2) > 20;
```
