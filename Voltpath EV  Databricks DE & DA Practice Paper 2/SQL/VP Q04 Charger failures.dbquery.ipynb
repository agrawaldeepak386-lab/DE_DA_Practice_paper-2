```sql
WITH charger_stats AS (
  SELECT
    charger_id,
    COUNT_IF(status = 'COMPLETED') AS completed_sessions,
    COUNT_IF(status = 'FAILED') AS failed_sessions
  FROM gold.fact_sessions
  GROUP BY charger_id
)
SELECT
  charger_id,
  completed_sessions,
  failed_sessions,
  ROUND(100.0 * failed_sessions / (completed_sessions + failed_sessions), 2) AS failure_rate_pct
FROM charger_stats
WHERE completed_sessions + failed_sessions >= 10
ORDER BY failure_rate_pct DESC, charger_id
LIMIT 10;
```
